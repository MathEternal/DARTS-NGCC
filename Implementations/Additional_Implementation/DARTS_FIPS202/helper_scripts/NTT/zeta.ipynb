```python
def bit_reverse(x: int, bits: int) -> int:
    """
    使用位操作将 x 的 bits 位做 bit-reversal
    例如： bit_reverse(3, 5) 把二进制 00011 -> 11000 
    """
    rev = 0
    for _ in range(bits):
        rev = (rev << 1) | (x & 1)
        x >>= 1
    return rev
```

```python
# 128 bits 和 256 bits 安全性

# Sign 中的 zetas 和 zetas_inv 的计算

# (ζ ^ bitrev(i)) * R mod q
# 这里乘 R 是因为 Montgomery 约简时会消去这一部分

q = 130817  # 模数
n = 512     # 多项式的次数
zeta = 667  # ζ 为模 q 下的 256 次本原单位根
R = 2 ** 32 # Montgomery 参数 
            # MONT = R mod q = 114369

sign_zetas = list()
sign_zetas_inv = list()

for i in range(128):
    sign_zetas.append(zeta**bit_reverse(i,7) * R % q)
    sign_zetas_inv.append(pow(zeta,-bit_reverse(i,7)-1,q) * R % q)

sign_zetas_inv[-1] = R ** 2 * pow(128,-1,q) % q

print(sign_zetas)
print(sign_zetas_inv)
```
