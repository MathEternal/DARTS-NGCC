```python
import numpy as np

# 计算 n = 512 时 FFT 所需的根
N = 512
scale = 65536
rootsf = np.exp(-1j * np.pi * np.arange(N) / N)
roots = [(int(r.real), int(r.imag)) for r in np.round(scale * rootsf)]

print("FFT 512 点所需的根（定点表示，实部和虚部均乘以 65536）：")
for i in range(0, N, 8):
    line = ', '.join([f"{{ .real = {r[0]:>6}, .imag = {r[1]:>6} }}" for r in roots[i:i+8]])
    print(line + ',')
```
